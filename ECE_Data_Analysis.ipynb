## Advantages of numpy arrays over lists

Python numpy arrays are more compact as compared to lists.

```python
# consider a list of numbers
num_list = [4,5,6,1,2,3]

# add 1 to each element
num_modified_list = [i + 1 for i in num_list]

# print the list
num_modified_list
```

```python
# create an array from a list
num_array = np.array([4,5,6,1,2,3])

# add 1 to each element
num_modified_array = num_array + 1

# print the array
num_modified_array
```

Python numpy array are fast as compared to lists

```python
# create a Numpy array using arange of size 1000000
my_array = np.arange(1000000)
print(my_array)
```

```python
# create a python list using range of size 1000000
my_list = list(range(1000000))
print(my_list)
```

## Pandas

**Why Pandas?** While NumPy handles raw numeric arrays, real-world data (lab measurement logs, sensor CSVs, student records) comes in **tables** with labeled columns and mixed data types. Pandas is built for exactly this.

### 2.1 Series — a labeled 1D array


```python
import pandas as pd
```

```python
temps = pd.Series([25.4, 26.1, 24.8, 27.3, 25.9],
                   index=['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
                   name='Temperature (C)')
print(temps)
print("\nAccess by label 'Wed':", temps['Wed'])
print("Mean temperature     :", temps.mean())
```

### 2.2 DataFrame — a labeled 2D table

Think of a DataFrame like an Excel sheet or a database table.


```python
# Simulated lab dataset: student circuit-lab measurements
data = {
    'StudentID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Resistor_Ohm': [220, 330, 220, 470, 330, 220, 470, 330],
    'Voltage_V': [4.98, 5.10, 4.85, 5.25, 5.05, 4.90, np.nan, 5.15],
    'Current_mA': [22.6, 15.4, 22.0, 11.1, 15.3, 22.3, 10.6, 15.6],
    'Branch': ['ECE', 'ECE', 'EEE', 'ECE', 'EEE', 'ECE', 'ECE', 'EEE']
}
df = pd.DataFrame(data)
df
```

```python
# Quick inspection tools
print(df.shape)        # (rows, columns)
print(df.dtypes)        # data types
df.head(3)               # first 3 rows
```

```python
df.info()          # summary: non-null counts, dtypes
```

```python
df.describe()      # statistical summary of numeric columns
```

### 2.3 Selecting & Filtering Data


```python
# Select a single column (returns a Series)
print(df['Voltage_V'])

# Select multiple columns
df[['StudentID', 'Voltage_V']]
```

```python
# Filtering rows with conditions (boolean masking, just like NumPy!)
ece_students = df[df['Branch'] == 'ECE']
print("ECE students only:")
ece_students
```

```python
# Combine conditions with & (and), | (or) -- note the parentheses!
high_voltage_ece = df[(df['Branch'] == 'ECE') & (df['Voltage_V'] > 5.0)]
high_voltage_ece
```

```python
# .loc (label-based) and .iloc (position-based) selection
print("Row at position 0:\n", df.iloc[0])
print("\nVoltage of row where StudentID == 103:")
print(df.loc[df['StudentID'] == 103, 'Voltage_V'])
```

### 2.4 Handling Missing Data

Notice student 107 has a missing (`NaN`) Voltage reading — very common with real sensor logs (dropped samples, faulty ADC reads, etc).


```python
print("Missing values per column:")
print(df.isnull().sum())

# Option 1: Fill missing value with the column mean
df_filled = df.copy()
df_filled['Voltage_V'] = df_filled['Voltage_V'].fillna(df_filled['Voltage_V'].mean())

# Option 2: Drop rows with any missing values
df_dropped = df.dropna()

print("\nFilled version (row 6):\n", df_filled.loc[6])
print("\nShape after dropna():", df_dropped.shape, " (original was", df.shape, ")")
```

### 2.5 Creating New Columns & Applying Functions

Ohm's law check: is measured resistance consistent with V = I * R?


```python
df_filled['Computed_R_Ohm'] = (df_filled['Voltage_V'] / (df_filled['Current_mA'] / 1000)).round(1)
df_filled['R_Error_%'] = (
    (df_filled['Computed_R_Ohm'] - df_filled['Resistor_Ohm']).abs()
    / df_filled['Resistor_Ohm'] * 100
).round(2)
df_filled
```

### 2.6 GroupBy — Aggregating Data by Category

Very useful for comparing groups (e.g., branch-wise, batch-wise, resistor-wise averages).


```python
branch_summary = df_filled.groupby('Branch')[['Voltage_V', 'Current_mA']].mean().round(3)
print("Average readings by branch:")
branch_summary
```

```python
resistor_summary = df_filled.groupby('Resistor_Ohm').agg(
    avg_voltage=('Voltage_V', 'mean'),
    avg_current=('Current_mA', 'mean'),
    count=('StudentID', 'count')
).round(3)
resistor_summary
```

### 2.7 Sorting & Value Counts


```python
print("Sorted by Voltage (descending):")
print(df_filled.sort_values('Voltage_V', ascending=False)[['StudentID', 'Voltage_V']])

print("\nHow many students per branch:")
print(df_filled['Branch'].value_counts())
```


# Part 3 — Matplotlib

**Why Matplotlib?** Plotting is essential for signal inspection (time-domain waveform, spectrum), debugging measurements, and reporting results.

### 3.1 Basic Line Plot


```python
import matplotlib.pyplot as plt
```

```python
t = np.linspace(0, 1, 500)
signal = np.sin(2 * np.pi * 5 * t)   # 5 Hz sine wave

plt.figure(figsize=(8, 4))
plt.plot(t, signal, color='navy', linewidth=1.5)
plt.title('5 Hz Sine Wave')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid(True)
plt.show()
```

### 3.2 Multiple Signals on One Plot


```python
sig_5hz = np.sin(2 * np.pi * 5 * t)
sig_10hz = np.sin(2 * np.pi * 10 * t)

plt.figure(figsize=(8, 4))
plt.plot(t, sig_5hz, label='5 Hz', color='steelblue')
plt.plot(t, sig_10hz, label='10 Hz', color='orangered', linestyle='--')
plt.title('Comparing Two Frequencies')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.show()
```

### 3.3 Scatter Plot — Voltage vs Current (from our lab DataFrame)


```python
plt.figure(figsize=(6, 5))
plt.scatter(df_filled['Current_mA'], df_filled['Voltage_V'],
            c='darkgreen', s=80, edgecolor='black')
plt.title('Voltage vs Current (Lab Data)')
plt.xlabel('Current (mA)')
plt.ylabel('Voltage (V)')
plt.grid(True)
plt.show()
```

### 3.4 Bar Chart — Average Voltage per Branch


```python
avg_v = df_filled.groupby('Branch')['Voltage_V'].mean()

plt.figure(figsize=(6, 4))
plt.bar(avg_v.index, avg_v.values, color=['#4C72B0', '#DD8452'])
plt.title('Average Voltage by Branch')
plt.xlabel('Branch')
plt.ylabel('Average Voltage (V)')
plt.show()
```

### 3.5 Histogram — Distribution of Noise

Useful for checking if sensor noise looks Gaussian.


```python
noise_samples = np.random.normal(0, 1, 1000)

plt.figure(figsize=(6, 4))
plt.hist(noise_samples, bins=30, color='slateblue', edgecolor='white')
plt.title('Distribution of 1000 Noise Samples')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

### 3.6 Subplots — Multiple Plots in a Grid

Very handy for comparing a clean signal, its noisy version, and the noise itself.


```python
fs = 500
t = np.linspace(0, 1, fs)
clean = np.sin(2 * np.pi * 5 * t)
noise = np.random.normal(0, 0.3, fs)
noisy = clean + noise

fig, axes = plt.subplots(3, 1, figsize=(8, 8), sharex=True)

axes[0].plot(t, clean, color='green')
axes[0].set_title('Clean Signal')

axes[1].plot(t, noise, color='gray')
axes[1].set_title('Noise')

axes[2].plot(t, noisy, color='crimson')
axes[2].set_title('Noisy Signal (Clean + Noise)')
axes[2].set_xlabel('Time (s)')

plt.tight_layout()
plt.show()
```

# Part 4 — Seaborn (35 min)

**Why Seaborn?** Built on top of Matplotlib, Seaborn makes **statistical plots** (distributions, correlations, category comparisons) much easier and prettier, and it works directly with Pandas DataFrames.

### 4.1 Building a Larger Sample Dataset

Let's simulate a bigger dataset (30 students) so our statistical plots are more meaningful.


```python
import seaborn as sns
```

```python
np.random.seed(1)
n = 30
branches = np.random.choice(['ECE', 'EEE', 'CSE'], size=n, p=[0.5, 0.3, 0.2])
resistors = np.random.choice([220, 330, 470], size=n)
voltage = 5 + np.random.normal(0, 0.15, n)
current = (voltage / resistors) * 1000 + np.random.normal(0, 0.5, n)  # mA

lab_df = pd.DataFrame({
    'Branch': branches,
    'Resistor_Ohm': resistors,
    'Voltage_V': voltage.round(3),
    'Current_mA': current.round(3)
})
lab_df['Power_mW'] = (lab_df['Voltage_V'] * lab_df['Current_mA']).round(2)
lab_df.head()
```

### 4.2 Distribution Plot — `histplot` / `kdeplot`


```python
plt.figure(figsize=(7, 4))
sns.histplot(lab_df['Voltage_V'], kde=True, color='teal')
plt.title('Distribution of Voltage Readings')
plt.show()
```

### 4.3 Box Plot — Compare Distributions Across Categories

Great for spotting outliers and comparing spread between branches.


```python
plt.figure(figsize=(7, 4))
sns.boxplot(data=lab_df, x='Branch', y='Current_mA', hue='Branch',
            palette='Set2', legend=False)
plt.title('Current Distribution by Branch')
plt.show()
```

### 4.4 Scatter Plot with Regression — `regplot` / `scatterplot`


```python
plt.figure(figsize=(6, 5))
sns.scatterplot(data=lab_df, x='Current_mA', y='Voltage_V',
                 hue='Branch', style='Branch', s=90)
plt.title('Voltage vs Current (colored by Branch)')
plt.show()
```

```python
plt.figure(figsize=(6, 5))
sns.regplot(data=lab_df, x='Resistor_Ohm', y='Current_mA',
            scatter_kws={'color': 'darkorange'}, line_kws={'color': 'navy'})
plt.title('Current vs Resistance (with trend line)')
plt.show()
```

### 4.5 Correlation Heatmap

Shows how strongly numeric columns relate to each other — great for quick sanity checks on multi-sensor datasets.


```python
numeric_cols = lab_df[['Resistor_Ohm', 'Voltage_V', 'Current_mA', 'Power_mW']]
corr = numeric_cols.corr()

plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', square=True)
plt.title('Correlation Heatmap')
plt.show()
```

### 4.6 Pair Plot — All Relationships at Once


```python
sns.pairplot(lab_df, vars=['Resistor_Ohm', 'Voltage_V', 'Current_mA'],
             hue='Branch', palette='husl', height=2.2)
plt.suptitle('Pairwise Relationships', y=1.02)
plt.show()
```

### 4.7 Bar Plot with Confidence Intervals


```python
plt.figure(figsize=(6, 4))
sns.barplot(data=lab_df, x='Branch', y='Power_mW', hue='Branch',
            palette='pastel', legend=False, errorbar='sd')
plt.title('Average Power by Branch (error bars = std dev)')
plt.show()
```
